```python
# create a simple if elif else code to check marks and give grade A , B, C, D, F based on the marks entered by the user.
marks = float(input("Enter your marks: "))

if 90 <= marks <= 100:
    grade = "A"
elif 80 <= marks < 90:
    grade = "B"
elif 70 <= marks < 80:
    grade = "C"
elif 60 <= marks < 70:
    grade = "D"
elif 0 <= marks < 60:
    grade = "F"
else:
    grade = "Invalid marks. Enter a value between 0 and 100."

print(f"Your grade is: {grade}")
```
